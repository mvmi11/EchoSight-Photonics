```python
!pip install -q trimesh plotly

import trimesh
import numpy as np
import plotly.graph_objects as go


def create_trace(mesh: trimesh.Trimesh, name: str, color: str, opacity: float = 1.0):
    x, y, z = mesh.vertices.T
    i, j, k = mesh.faces.T
    return go.Mesh3d(
        x=x, y=y, z=z, i=i, j=j, k=k,
        name=name, color=color, opacity=opacity,
        flatshading=False,
        lighting=dict(ambient=0.5, diffuse=0.8, roughness=0.3, specular=0.5)
    )


def build_3d_flying_wing_uav():
    traces = []
    solid_parts = []

    # 1. Center Fuselage & Swept Wings (Semi-Transparent Aerodynamic Shell)
    # Center body
    fuselage = trimesh.creation.box(extents=(70.0, 140.0, 18.0))
    fuselage.apply_translation((0.0, 0.0, 10.0))
    traces.append(create_trace(fuselage, "Center Fuselage (Shell)", "#CFD8DC", opacity=0.35))
    solid_parts.append(fuselage)

    # Left & Right Swept Wings
    for side in (-1.0, 1.0):
        wing = trimesh.creation.box(extents=(160.0, 65.0, 8.0))
        sweep = trimesh.transformations.rotation_matrix(np.radians(side * -22.0), (0.0, 0.0, 1.0))
        dihedral = trimesh.transformations.rotation_matrix(np.radians(side * 4.0), (0.0, 1.0, 0.0))
        wing.apply_transform(sweep @ dihedral)
        wing.apply_translation((side * 105.0, -22.0, 11.0))
        traces.append(create_trace(wing, f"{'Right' if side > 0 else 'Left'} Swept Wing", "#B0BEC5", opacity=0.45))
        solid_parts.append(wing)

        # Vertical Stabilizer Fins (under/over wing)
        fin = trimesh.creation.box(extents=(3.0, 35.0, 22.0))
        fin.apply_translation((side * 42.0, -52.0, 6.0))
        traces.append(create_trace(fin, "Vertical Stabilizer", "#455A64", opacity=0.9))
        solid_parts.append(fin)

    # 2. Front Puller Brushless Motor & Propeller (+Y Nose)
    rot_x = trimesh.transformations.rotation_matrix(np.pi / 2.0, (1.0, 0.0, 0.0))
    motor_bell = trimesh.creation.cylinder(radius=11.0, height=18.0, sections=36)
    motor_bell.apply_transform(rot_x)
    motor_bell.apply_translation((0.0, 78.0, 12.0))
    traces.append(create_trace(motor_bell, "Outrunner Brushless Motor", "#D4AF37", opacity=1.0))
    solid_parts.append(motor_bell)

    propeller = trimesh.creation.box(extents=(95.0, 3.0, 8.0))
    propeller.apply_translation((0.0, 89.0, 12.0))
    traces.append(create_trace(propeller, "Puller Propeller", "#212121", opacity=1.0))
    solid_parts.append(propeller)

    # 3. Bottom EO/IR Gimbal Camera Ball (Underbelly)
    gimbal_ball = trimesh.creation.icosphere(subdivisions=3, radius=18.0)
    gimbal_ball.apply_translation((0.0, 15.0, -8.0))
    traces.append(create_trace(gimbal_ball, "EO/IR Gimbal Turret", "#263238", opacity=0.85))
    solid_parts.append(gimbal_ball)

    cam_lens = trimesh.creation.cylinder(radius=7.0, height=10.0, sections=32)
    cam_lens.apply_transform(rot_x)
    cam_lens.apply_translation((0.0, 31.0, -8.0))
    traces.append(create_trace(cam_lens, "Optical Camera Lens", "#00BCD4", opacity=1.0))
    solid_parts.append(cam_lens)

    # 4. Internal Electronics (Visible Through the Transparent Shell!)
    # A. Flight Controller & Optical PCB (Green)
    pcb = trimesh.creation.box(extents=(44.0, 65.0, 3.0))
    pcb.apply_translation((0.0, 10.0, 12.0))
    traces.append(create_trace(pcb, "OpticDroneUnit PCB", "#2E7D32", opacity=1.0))
    solid_parts.append(pcb)

    # B. 4S Li-Po Battery Pack (Blue)
    battery = trimesh.creation.box(extents=(38.0, 45.0, 10.0))
    battery.apply_translation((0.0, 35.0, 11.0))
    traces.append(create_trace(battery, "4S Li-Po Battery", "#1565C0", opacity=1.0))
    solid_parts.append(battery)

    # C. SFP Optical Transceiver & Rear Fiber Spool (Silver & Yellow)
    sfp = trimesh.creation.box(extents=(14.0, 22.0, 8.0))
    sfp.apply_translation((0.0, -28.0, 12.0))
    traces.append(create_trace(sfp, "SFP Optical Module", "#90A4AE", opacity=1.0))
    solid_parts.append(sfp)

    spool = trimesh.creation.cylinder(radius=16.0, height=24.0, sections=36)
    spool.apply_transform(rot_x)
    spool.apply_translation((0.0, -52.0, 11.0))
    traces.append(create_trace(spool, "G.657.A2 Fiber Spool", "#FBC02D", opacity=1.0))
    solid_parts.append(spool)

    # Export printable STL
    uav_mesh = trimesh.util.concatenate(solid_parts)
    uav_mesh.export("FlyingWing_3D_UAV.stl")

    # Render Interactive 3D Scene
    fig = go.Figure(data=traces)
    fig.update_layout(
        title="Flying-Wing Fiber-Optic UAV: 3D External & Internal CAD Model",
        scene=dict(
            xaxis_title="Wingspan X (mm)",
            yaxis_title="Length Y (mm)",
            zaxis_title="Height Z (mm)",
            aspectmode="data"
        ),
        margin=dict(l=0, r=0, b=0, t=40),
        legend=dict(x=0.01, y=0.95)
    )
    fig.show()


build_3d_flying_wing_uav()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches


class EmbeddedMCUSimulator:
    """
    Simulates the onboard Microcontroller Unit (MCU) hardware interfaces:
    - 12-bit ADC acoustic quantization (0 - 4095 levels)
    - Closed-loop PID to 50Hz Servo PWM pulse-width translation (1000us - 2000us)
    - SPI high-speed camera frame transfer timing
    """

    def __init__(self, adc_bits: int = 12, pwm_freq_hz: float = 50.0, spi_clock_mhz: float = 20.0):
        self.adc_levels = (2 ** adc_bits) - 1
        self.pwm_freq = pwm_freq_hz
        self.pwm_period_ms = 1000.0 / pwm_freq_hz
        self.spi_clock_mhz = spi_clock_mhz

    def simulate_12bit_adc(self, duration_ms: float = 15.0, fs: int = 44100):
        t = np.linspace(0, duration_ms / 1000.0, int(fs * (duration_ms / 1000.0)))
        # Continuous analog mic voltage (0.0V to 3.3V centered at 1.65V)
        analog_v = 1.65 + 0.9 * np.sin(2 * np.pi * 240.0 * t) + 0.25 * np.random.normal(0, 1, len(t))
        analog_v = np.clip(analog_v, 0.0, 3.3)
        # Quantize into 12-bit discrete digital integers (0..4095)
        digital_adc = np.round((analog_v / 3.3) * self.adc_levels).astype(int)
        return t * 1000.0, analog_v, digital_adc

    def angle_to_pwm_us(self, angle_deg: float) -> float:
        """Maps gimbal angle (-45 to +45 deg) to standard servo PWM pulse (1000us to 2000us)."""
        center_us = 1500.0
        us_per_deg = 500.0 / 45.0
        return round(center_us + (angle_deg * us_per_deg), 1)

    def generate_pwm_waveform(self, pulse_us: float, duration_ms: float = 40.0):
        t_ms = np.linspace(0, duration_ms, 2000)
        cycle_pos_ms = np.mod(t_ms, self.pwm_period_ms)
        pulse_ms = pulse_us / 1000.0
        pwm_voltage = np.where(cycle_pos_ms <= pulse_ms, 3.3, 0.0)
        return t_ms, pwm_voltage

    def compute_spi_latency_ms(self, width: int = 480, height: int = 360) -> float:
        total_bits = width * height * 8
        transfer_sec = total_bits / (self.spi_clock_mhz * 1e6)
        return round(transfer_sec * 1000.0, 2)

    def render_embedded_dashboard(self, pan_deg: float = -1.2, tilt_deg: float = -0.9):
        t_adc, analog_v, digital_adc = self.simulate_12bit_adc()
        pan_pwm_us = self.angle_to_pwm_us(pan_deg)
        tilt_pwm_us = self.angle_to_pwm_us(tilt_deg)
        spi_ms = self.compute_spi_latency_ms()

        t_pwm, pan_wave = self.generate_pwm_waveform(pan_pwm_us)
        _, tilt_wave = self.generate_pwm_waveform(tilt_pwm_us)

        fig = plt.figure(figsize=(15, 9), dpi=200, facecolor="#0F172A")
        gs = fig.add_gridspec(2, 2, height_ratios=[1.1, 1.0])
        fig.suptitle("EchoSight-Photonics: Embedded MCU Architecture & Hardware Signal Verification",
                     color="white", fontsize=14, fontweight="bold", y=0.96)

        # --- TOP PANEL: EMBEDDED HARDWARE BLOCK DIAGRAM ---
        ax_arch = fig.add_subplot(gs[0, :], facecolor="#1E293B")
        ax_arch.set_xlim(0, 100)
        ax_arch.set_ylim(0, 40)
        ax_arch.axis("off")
        ax_arch.set_title("1. Onboard Flight Controller & Sensor Interface Architecture (SPI / ADC / PWM)",
                          color="#38BDF8", fontweight="bold", pad=8)

        def draw_block(x, y, w, h, title, subtitle, edge_col, fill_col="#0F172A"):
            rect = patches.FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.6",
                                          linewidth=2, edgecolor=edge_col, facecolor=fill_col)
            ax_arch.add_patch(rect)
            ax_arch.text(x + w/2, y + h*0.64, title, color="white", ha="center", va="center", fontweight="bold", fontsize=10)
            ax_arch.text(x + w/2, y + h*0.28, subtitle, color="#94A3B8", ha="center", va="center", fontsize=8.5)

        # Peripheral Input Blocks (Left)
        draw_block(3, 24, 22, 11, "EO/IR Optical Camera", "480x360 Grayscale Frame", "#00E5FF")
        draw_block(3, 5, 22, 11, "4-Mic Acoustic Array", "Analog Rotor Harmonics", "#FACC15")

        # Central MCU Block (Center)
        draw_block(38, 8, 24, 24, "CENTRAL MCU / AUTOPILOT",
                   "Kalman + Hough + FFT + PID\nReal-Time Embedded Core", "#22C55E", "#14281D")

        # Actuator & Optical Output Blocks (Right)
        draw_block(75, 24, 22, 11, "FSO Gimbal Actuators", f"Pan: {pan_deg}° | Tilt: {tilt_deg}°", "#F43F5E")
        draw_block(75, 5, 22, 11, "1550nm FSO & SFP Link", "160 Gbps WDM / Laser Lock", "#A855F7")

        # Protocol Arrows
        arrow_kw = dict(arrowstyle="->,head_width=0.35,head_length=0.45", lw=2.2)
        ax_arch.annotate("", xy=(37.5, 29), xytext=(25.5, 29), arrowprops=dict(color="#00E5FF", **arrow_kw))
        ax_arch.text(31.5, 31.2, f"SPI Bus ({self.spi_clock_mhz} MHz)\nLatency: {spi_ms} ms", color="#00E5FF", ha="center", fontsize=8, fontweight="bold")

        ax_arch.annotate("", xy=(37.5, 11), xytext=(25.5, 11), arrowprops=dict(color="#FACC15", **arrow_kw))
        ax_arch.text(31.5, 13.2, "12-Bit ADC\n(44.1 kHz / 4096 lvl)", color="#FACC15", ha="center", fontsize=8, fontweight="bold")

        ax_arch.annotate("", xy=(74.5, 29), xytext=(62.5, 29), arrowprops=dict(color="#F43F5E", **arrow_kw))
        ax_arch.text(68.5, 31.2, f"50Hz PWM Signals\n{pan_pwm_us}us / {tilt_pwm_us}us", color="#F43F5E", ha="center", fontsize=8, fontweight="bold")

        ax_arch.annotate("", xy=(74.5, 11), xytext=(62.5, 11), arrowprops=dict(color="#A855F7", **arrow_kw))
        ax_arch.text(68.5, 13.2, "UART / Optical\nTelemetry Lock", color="#A855F7", ha="center", fontsize=8, fontweight="bold")

        # --- BOTTOM LEFT: 12-BIT ADC ACOUSTIC SAMPLING ---
        ax_adc = fig.add_subplot(gs[1, 0], facecolor="#1E293B")
        ax_adc.step(t_adc[:180], digital_adc[:180], where="mid", color="#FACC15", linewidth=1.3, label="12-Bit ADC Digital Samples")
        ax_adc.set_title("2. ADC Acoustic Digitization (240 Hz Rotor Signal)", color="white", fontweight="bold")
        ax_adc.set_xlabel("Time (ms)", color="white")
        ax_adc.set_ylabel("ADC Digital Code (0 - 4095)", color="white")
        ax_adc.tick_params(colors="white")
        ax_adc.grid(True, color="#334155", linestyle="--", alpha=0.6)
        ax_adc.legend(facecolor="#0F172A", labelcolor="white", fontsize=8.5)

        # --- BOTTOM RIGHT: PWM SERVO ACTUATOR SIGNALS ---
        ax_pwm = fig.add_subplot(gs[1, 1], facecolor="#1E293B")
        ax_pwm.plot(t_pwm, pan_wave, color="#F43F5E", linewidth=2, label=f"Pan PWM ({pan_deg}° -> {pan_pwm_us} us)")
        ax_pwm.plot(t_pwm, tilt_wave - 4.0, color="#38BDF8", linewidth=2, label=f"Tilt PWM ({tilt_deg}° -> {tilt_pwm_us} us)")
        ax_pwm.set_title("3. MCU PWM Actuator Control Waveforms (50 Hz)", color="white", fontweight="bold")
        ax_pwm.set_xlabel("Time (ms)", color="white")
        ax_pwm.set_ylabel("Logic Voltage (3.3V TTL)", color="white")
        ax_pwm.set_yticks([0, 3.3, -4.0, -0.7])
        ax_pwm.set_yticklabels(["0V", "3.3V", "0V", "3.3V"])
        ax_pwm.tick_params(colors="white")
        ax_pwm.grid(True, color="#334155", linestyle="--", alpha=0.6)
        ax_pwm.legend(loc="upper right", facecolor="#0F172A", labelcolor="white", fontsize=8.5)

        plt.tight_layout(rect=(0, 0.02, 1, 0.93))
        plt.savefig("EchoSight_Embedded_MCU_Architecture.png", dpi=200, facecolor="#0F172A")
        plt.show()


# --- Execute Embedded Hardware Simulation ---
mcu_sim = EmbeddedMCUSimulator()
mcu_sim.render_embedded_dashboard(pan_deg=-1.2, tilt_deg=-0.9)
```

```python
!pip install -q imageio[ffmpeg]

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from IPython.display import HTML, display


def transform_pts(pts, pos, roll, pitch, yaw):
    """Applies 3D Roll, Pitch, Yaw rotation and translation to local coordinates."""
    x0, y0, z0 = pos
    cr, sr = np.cos(roll), np.sin(roll)
    cp, sp = np.cos(pitch), np.sin(pitch)
    cy, sy = np.cos(yaw), np.sin(yaw)

    Rx = np.array([[1, 0, 0], [0, cr, -sr], [0, sr, cr]])
    Ry = np.array([[cp, 0, sp], [0, 1, 0], [-sp, 0, cp]])
    Rz = np.array([[cy, -sy, 0], [sy, cy, 0], [0, 0, 1]])
    R = Rz @ Ry @ Rx

    arr = np.array(pts, dtype=float)
    rotated = (arr @ R.T) + np.array([x0, y0, z0])
    return rotated.tolist()


def build_quadcopter_geometry(pos, roll, pitch, yaw, rotor_angle):
    """Builds 3D geometry for the X-frame Quadcopter, Landing Skids, 4 Rotors, and 3-Axis Gimbal."""
    # 1. Central Avionics Hub (Top Plate & Bottom Plate)
    hub_top = transform_pts([(-4, -5, 1.5), (4, -5, 1.5), (4, 5, 1.5), (-4, 5, 1.5)], pos, roll, pitch, yaw)
    hub_bot = transform_pts([(-4, -5, -1.5), (4, -5, -1.5), (4, 5, -1.5), (-4, 5, -1.5)], pos, roll, pitch, yaw)

    # 2. Four X-Frame Arms & Spinning Rotor Blades
    arm_tips_local = [(14, 14, 1.0), (-14, 14, 1.0), (-14, -14, 1.0), (14, -14, 1.0)]
    arms = []
    rotors = []
    for idx, (ax_l, ay_l, az_l) in enumerate(arm_tips_local):
        p_start = transform_pts([(0, 0, 0.5)], pos, roll, pitch, yaw).pop()
        p_end = transform_pts([(ax_l, ay_l, az_l)], pos, roll, pitch, yaw).pop()
        arms.append((p_start, p_end))

        # Spinning blade line at each motor tip
        direction = 1.0 if idx % 2 == 0 else -1.0
        ang = rotor_angle * direction + (idx * np.pi / 4.0)
        bx = 7.5 * np.cos(ang)
        by = 7.5 * np.sin(ang)
        b1 = transform_pts([(ax_l - bx, ay_l - by, az_l + 0.8)], pos, roll, pitch, yaw).pop()
        b2 = transform_pts([(ax_l + bx, ay_l + by, az_l + 0.8)], pos, roll, pitch, yaw).pop()
        rotors.append((b1, b2))

    # 3. Tall Landing Gear Skids (Matching Bottom Reference Image)
    skids = []
    for side in (-6.5, 6.5):
        strut_f = transform_pts([(side * 0.6, 4.0, -1.5), (side, 5.5, -9.5)], pos, roll, pitch, yaw)
        strut_b = transform_pts([(side * 0.6, -4.0, -1.5), (side, -5.5, -9.5)], pos, roll, pitch, yaw)
        rail = transform_pts([(side, -10.0, -9.5), (side, 10.0, -9.5)], pos, roll, pitch, yaw)
        skids.extend([strut_f, strut_b, rail])

    # 4. Suspended 3-Axis Stabilized Gimbal Camera (Stabilized horizontal despite drone tilt!)
    gimbal_center = transform_pts([(0.0, 2.5, -4.8)], pos, roll, pitch, yaw).pop()
    cam_box = transform_pts([
        (-2.8, 0.5, -6.5), (2.8, 0.5, -6.5), (2.8, 4.5, -6.5), (-2.8, 4.5, -6.5)
    ], pos, roll * 0.15, pitch * 0.15, yaw)

    fiber_spool_pt = transform_pts([(0.0, -5.0, -1.5)], pos, roll, pitch, yaw).pop()
    return [hub_top, hub_bot], [cam_box], arms, rotors, skids, gimbal_center, fiber_spool_pt


# --- Setup 3D Airspace Arena ---
fig = plt.figure(figsize=(12, 8), dpi=130, facecolor="#0B131E")
ax = fig.add_subplot(111, projection="3d", facecolor="#0B131E")

gx = np.linspace(-100, 100, 12)
gy = np.linspace(-100, 100, 12)
GX, GY = np.meshgrid(gx, gy)
ax.plot_wireframe(GX, GY, np.zeros_like(GX), color="#1E3A5F", linewidth=0.6, alpha=0.5)

# EchoSight Ground Station at (0, 0, 0)
ax.scatter([0], [0], [2], color="#00E676", s=130, marker="^", label="EchoSight Ground Station (PID + SFP + 4-Mic)")

# Flight Trajectory (75 Frames)
frames = 75
t = np.linspace(0, 1, frames)
traj_x = -70.0 + 140.0 * t
traj_y = -60.0 + 130.0 * t + 16.0 * np.sin(np.pi * t)
traj_z = 38.0 + 20.0 * np.sin(np.pi * t * 0.85)

# Plot Artists
body_poly = Poly3DCollection([], facecolors=["#263238", "#37474F"], edgecolors="#90A4AE", linewidths=0.6)
gimbal_poly = Poly3DCollection([], facecolors=["#00E5FF"], edgecolors="#006064", linewidths=0.8)
ax.add_collection3d(body_poly)
ax.add_collection3d(gimbal_poly)

arm_lines = [ax.plot([], [], [], color="#CFD8DC", linewidth=2.5)[0] for _ in range(4)]
rotor_lines = [ax.plot([], [], [], color="#FFEA00", linewidth=2.0)[0] for _ in range(4)]
skid_lines = [ax.plot([], [], [], color="#78909C", linewidth=1.6)[0] for _ in range(6)]

fso_laser_line, = ax.plot([], [], [], color="#FF1744", linewidth=2.2, label="1550nm FSO Laser Lock (3-Axis Gimbal)")
fiber_tether_line, = ax.plot([], [], [], color="#FFD600", linewidth=1.5, label="G.657.A2 Optical Fiber Tether")
kalman_pred_line, = ax.plot([], [], [], color="#00E5FF", linewidth=2.4, linestyle="--", label="Kalman 3D Trajectory Prediction")
acoustic_beam_line, = ax.plot([], [], [], color="#76FF03", linewidth=1.5, linestyle=":", label="Acoustic TDOA Bearing Vector")

hud_text = ax.text2D(0.02, 0.93, "", transform=ax.transAxes, color="white", fontsize=9.5, family="monospace",
                     bbox=dict(boxstyle="round,pad=0.4", facecolor="#132238", edgecolor="#00E5FF", alpha=0.88))

ax.set_xlim((-100, 100))
ax.set_ylim((-100, 100))
ax.set_zlim((0, 80))
ax.set_xlabel("East-West X (m)", color="white")
ax.set_ylabel("North-South Y (m)", color="white")
ax.set_zlabel("Altitude Z (m)", color="white")
ax.tick_params(colors="white")
ax.set_title("EchoSight-Photonics: Quadcopter (3-Axis Gimbal) & Optical Tracking Simulation", color="white", fontweight="bold", pad=12)
ax.legend(loc="lower left", facecolor="#132238", edgecolor="#37474F", labelcolor="white", fontsize=8.5)


def update(frame):
    pos = (float(traj_x[frame]), float(traj_y[frame]), float(traj_z[frame]))
    if frame < frames - 1:
        dx = float(traj_x[frame + 1] - traj_x[frame])
        dy = float(traj_y[frame + 1] - traj_y[frame])
    else:
        dx = float(traj_x[frame] - traj_x[frame - 1])
        dy = float(traj_y[frame] - traj_y[frame - 1])

    yaw = np.arctan2(dy, dx) - np.pi / 2.0
    pitch = np.radians(-12.0)  # Forward flight pitch tilt
    roll = np.radians(8.0 * np.sin(frame * 0.15))  # Dynamic banking roll

    hub_verts, cam_verts, arms, rotors, skids, gimbal_pt, spool_pt = build_quadcopter_geometry(
        pos, roll, pitch, yaw, frame * 1.1
    )

    body_poly.set_verts(hub_verts)
    gimbal_poly.set_verts(cam_verts)

    for idx in range(4):
        p1, p2 = arms[idx]
        arm_lines[idx].set_data([p1[0], p2[0]], [p1[1], p2[1]])
        arm_lines[idx].set_3d_properties([p1[2], p2[2]])

        r1, r2 = rotors[idx]
        rotor_lines[idx].set_data([r1[0], r2[0]], [r1[1], r2[1]])
        rotor_lines[idx].set_3d_properties([r1[2], r2[2]])

    for idx in range(6):
        s1, s2 = skids[idx]
        skid_lines[idx].set_data([s1[0], s2[0]], [s1[1], s2[1]])
        skid_lines[idx].set_3d_properties([s1[2], s2[2]])

    # 1550nm FSO Tracking Laser (Ground Station -> 3-Axis Gimbal)
    fso_laser_line.set_data([0.0, gimbal_pt[0]], [0.0, gimbal_pt[1]])
    fso_laser_line.set_3d_properties([2.0, gimbal_pt[2]])

    # Acoustic TDOA Vector
    acoustic_beam_line.set_data([0.0, pos[0]], [0.0, pos[1]])
    acoustic_beam_line.set_3d_properties([2.0, pos[2]])

    # Trailing G.657.A2 Optical Fiber Tether
    t_sag = np.linspace(0, 1, 20)
    fx = -85.0 + (spool_pt[0] + 85.0) * t_sag
    fy = -75.0 + (spool_pt[1] + 75.0) * t_sag
    fz = spool_pt[2] * (t_sag ** 1.6)
    fiber_tether_line.set_data(fx, fy)
    fiber_tether_line.set_3d_properties(fz)

    # Kalman 3D Future Path Prediction
    end_idx = min(frames, frame + 14)
    kalman_pred_line.set_data(traj_x[frame:end_idx], traj_y[frame:end_idx])
    kalman_pred_line.set_3d_properties(traj_z[frame:end_idx])

    azimuth = np.degrees(np.arctan2(pos[0], pos[1])) % 360.0
    dist_m = np.linalg.norm(pos)
    hud_text.set_text(
        f"PLATFORM     : X-QUADCOPTER + 3-AXIS DAMPED GIMBAL\n"
        f"ATTITUDE     : Roll={np.degrees(roll):+05.1f}° | Pitch={np.degrees(pitch):+05.1f}° | Yaw={np.degrees(yaw):+06.1f}°\n"
        f"ACOUSTIC TDOA: Azimuth={azimuth:06.2f}° | Rotor FFT=240.0 Hz\n"
        f"KALMAN & FSO : Range={dist_m:05.1f} m | Laser Lock Eff=96.4%"
    )

    ax.view_init(elev=22, azim=-55 + frame * 0.45)
    return [body_poly, gimbal_poly, fso_laser_line, fiber_tether_line, kalman_pred_line, acoustic_beam_line, hud_text]


anim = FuncAnimation(fig, update, frames=frames, interval=60, blit=False)
anim.save("EchoSight_Quadcopter_Gimbal.mp4", writer="ffmpeg", fps=20)
plt.close(fig)

print("Video generated successfully: EchoSight_Quadcopter_Gimbal.mp4")
display(HTML(anim.to_jshtml()))
```

```python
!pip install -q imageio[ffmpeg]

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from IPython.display import HTML, display


def rotate_translate(pts, pos, roll, pitch, yaw):
    x0, y0, z0 = pos
    cr, sr = np.cos(roll), np.sin(roll)
    cp, sp = np.cos(pitch), np.sin(pitch)
    cy, sy = np.cos(yaw), np.sin(yaw)

    Rx = np.array(((1.0, 0.0, 0.0), (0.0, cr, -sr), (0.0, sr, cr)), dtype=float)
    Ry = np.array(((cp, 0.0, sp), (0.0, 1.0, 0.0), (-sp, 0.0, cp)), dtype=float)
    Rz = np.array(((cy, -sy, 0.0), (sy, cy, 0.0), (0.0, 0.0, 1.0)), dtype=float)
    R = Rz @ Ry @ Rx

    arr = np.array(pts, dtype=float)
    return ((arr @ R.T) + np.array((x0, y0, z0))).tolist()


def build_drone_mesh(pos, roll, pitch, yaw, rotor_angle, scale=1.0):
    """Builds 3D geometry for an X-Frame Quadcopter with rotors and bottom sensor pod."""
    s = scale
    hub_top = rotate_translate([(-3.2*s, -4.0*s, 1.0*s), (3.2*s, -4.0*s, 1.0*s), (3.2*s, 4.0*s, 1.0*s), (-3.2*s, 4.0*s, 1.0*s)], pos, roll, pitch, yaw)
    hub_bot = rotate_translate([(-3.2*s, -4.0*s, -1.0*s), (3.2*s, -4.0*s, -1.0*s), (3.2*s, 4.0*s, -1.0*s), (-3.2*s, 4.0*s, -1.0*s)], pos, roll, pitch, yaw)

    arm_tips = ((11.0*s, 11.0*s, 0.7*s), (-11.0*s, 11.0*s, 0.7*s), (-11.0*s, -11.0*s, 0.7*s), (11.0*s, -11.0*s, 0.7*s))
    arms, rotors = [], []
    for idx, (ax_l, ay_l, az_l) in enumerate(arm_tips):
        p0 = rotate_translate([(0.0, 0.0, 0.3*s)], pos, roll, pitch, yaw).pop()
        p1 = rotate_translate([(ax_l, ay_l, az_l)], pos, roll, pitch, yaw).pop()
        arms.append((p0, p1))

        d = 1.0 if idx % 2 == 0 else -1.0
        ang = rotor_angle * d + (idx * np.pi / 4.0)
        bx, by = 6.0 * s * np.cos(ang), 6.0 * s * np.sin(ang)
        r0 = rotate_translate([(ax_l - bx, ay_l - by, az_l + 0.5*s)], pos, roll, pitch, yaw).pop()
        r1 = rotate_translate([(ax_l + bx, ay_l + by, az_l + 0.5*s)], pos, roll, pitch, yaw).pop()
        rotors.append((r0, r1))

    sensor_pt = rotate_translate([(0.0, 3.5*s, -1.8*s)], pos, roll, pitch, yaw).pop()
    tail_pt = rotate_translate([(0.0, -4.2*s, -1.0*s)], pos, roll, pitch, yaw).pop()
    return [hub_top, hub_bot], arms, rotors, sensor_pt, tail_pt


# --- Setup 3D Airspace Arena ---
fig = plt.figure(figsize=(13, 8.5), dpi=130, facecolor="#0B131E")
ax = fig.add_subplot(111, projection="3d", facecolor="#0B131E")

gx = np.linspace(-100, 100, 13)
gy = np.linspace(-100, 100, 13)
GX, GY = np.meshgrid(gx, gy)
ax.plot_wireframe(GX, GY, np.zeros_like(GX), color="#1E3A5F", linewidth=0.6, alpha=0.5)

frames = 85
t = np.linspace(0, 1, frames)

# Trajectory 1: Non-Cooperative Fiber-Optic Drone (Target)
t_x = -65.0 + 135.0 * t
t_y = -45.0 + 115.0 * t + 18.0 * np.sin(1.5 * np.pi * t)
t_z = 38.0 + 14.0 * np.sin(np.pi * t)

# Trajectory 2: EchoSight Interceptor/Tracker Drone (Manoeuvring to intercept & lock)
i_x = 75.0 - 35.0 * t + 85.0 * (t ** 1.4) - 45.0 * np.sin(np.pi * t)
i_y = -75.0 + 125.0 * t
i_z = 55.0 - 12.0 * t + 10.0 * np.sin(np.pi * t)

# Artists for Drone 1 (EchoSight Tracker UAV - Cyan/Green)
tracker_body = Poly3DCollection([], facecolors=("#00E676", "#00B0FF"), edgecolors="#E0F7FA", linewidths=0.6)
ax.add_collection3d(tracker_body)
tracker_arms = [ax.plot([], [], [], color="#00E5FF", linewidth=2.4).pop() for _ in range(4)]
tracker_rotors = [ax.plot([], [], [], color="#76FF03", linewidth=2.0).pop() for _ in range(4)]

# Artists for Drone 2 (Non-Cooperative Fiber UAV - Orange/Red)
target_body = Poly3DCollection([], facecolors=("#FF6D00", "#DD2C00"), edgecolors="#FFCCBC", linewidths=0.6)
ax.add_collection3d(target_body)
target_arms = [ax.plot([], [], [], color="#FF9E80", linewidth=2.4).pop() for _ in range(4)]
target_rotors = [ax.plot([], [], [], color="#FFD600", linewidth=2.0).pop() for _ in range(4)]

# Dynamic Links & Trajectories
fso_beam, = ax.plot([], [], [], color="#FF1744", linewidth=2.5, label="Air-to-Air 1550nm FSO Optical Lock")
fiber_tether, = ax.plot([], [], [], color="#FFD600", linewidth=1.4, label="Target G.657.A2 Optical Fiber Tether")
kalman_path, = ax.plot([], [], [], color="#00E5FF", linewidth=2.2, linestyle="--", label="Kalman 3D Intercept Prediction")
tracker_trail, = ax.plot([], [], [], color="#00E676", linewidth=1.4, linestyle=":", label="EchoSight Tracker Flight Path")

hud_text = ax.text2D(0.02, 0.91, "", transform=ax.transAxes, color="white", fontsize=9.5, family="monospace",
                     bbox=dict(boxstyle="round,pad=0.45", facecolor="#132238", edgecolor="#00E5FF", alpha=0.9))

ax.set_xlim((-100, 100))
ax.set_ylim((-100, 100))
ax.set_zlim((0, 80))
ax.set_xlabel("East-West X (m)", color="white")
ax.set_ylabel("North-South Y (m)", color="white")
ax.set_zlabel("Altitude Z (m)", color="white")
ax.tick_params(colors="white")
ax.set_title("EchoSight-Photonics: Air-to-Air Autonomous UAV Tracking & Optical Lock", color="white", fontweight="bold", pad=12)
ax.legend(loc="lower left", facecolor="#132238", edgecolor="#37474F", labelcolor="white", fontsize=8.5)


def get_heading(xs, ys, idx):
    if idx < len(xs) - 1:
        dx = float(xs[idx + 1] - xs[idx])
        dy = float(ys[idx + 1] - ys[idx])
    else:
        dx = float(xs[idx] - xs[idx - 1])
        dy = float(ys[idx] - ys[idx - 1])
    return np.arctan2(dy, dx) - np.pi / 2.0


def update(frame):
    # Target Drone State
    pos_t = (float(t_x[frame]), float(t_y[frame]), float(t_z[frame]))
    yaw_t = get_heading(t_x, t_y, frame)
    roll_t = np.radians(10.0 * np.cos(frame * 0.15))
    pitch_t = np.radians(-10.0)

    t_verts, t_arms, t_rots, t_sensor, t_tail = build_drone_mesh(pos_t, roll_t, pitch_t, yaw_t, frame * 1.2, scale=0.95)
    target_body.set_verts(t_verts)
    for k in range(4):
        p0, p1 = t_arms[k]
        x0, y0, z0 = p0
        x1, y1, z1 = p1
        target_arms[k].set_data((x0, x1), (y0, y1))
        target_arms[k].set_3d_properties((z0, z1))

        r0, r1 = t_rots[k]
        rx0, ry0, rz0 = r0
        rx1, ry1, rz1 = r1
        target_rotors[k].set_data((rx0, rx1), (ry0, ry1))
        target_rotors[k].set_3d_properties((rz0, rz1))

    # EchoSight Tracker Drone State (Banking towards Target)
    pos_i = (float(i_x[frame]), float(i_y[frame]), float(i_z[frame]))
    yaw_i = get_heading(i_x, i_y, frame)
    roll_i = np.radians(-14.0 * np.sin(frame * 0.12))
    pitch_i = np.radians(-12.0)

    i_verts, i_arms, i_rots, i_sensor, _ = build_drone_mesh(pos_i, roll_i, pitch_i, yaw_i, frame * 1.3, scale=1.05)
    tracker_body.set_verts(i_verts)
    for k in range(4):
        p0, p1 = i_arms[k]
        x0, y0, z0 = p0
        x1, y1, z1 = p1
        tracker_arms[k].set_data((x0, x1), (y0, y1))
        tracker_arms[k].set_3d_properties((z0, z1))

        r0, r1 = i_rots[k]
        rx0, ry0, rz0 = r0
        rx1, ry1, rz1 = r1
        tracker_rotors[k].set_data((rx0, rx1), (ry0, ry1))
        tracker_rotors[k].set_3d_properties((rz0, rz1))

    # Air-to-Air 1550nm FSO Tracking Laser (Tracker Sensor -> Target Drone)
    ix, iy, iz = i_sensor
    tx_s, ty_s, tz_s = t_sensor
    fso_beam.set_data((ix, tx_s), (iy, ty_s))
    fso_beam.set_3d_properties((iz, tz_s))

    # Trailing Optical Fiber Tether from Target Drone to Ground Spool
    tail_x, tail_y, tail_z = t_tail
    t_sag = np.linspace(0, 1, 20)
    fx = -85.0 + (tail_x + 85.0) * t_sag
    fy = -65.0 + (tail_y + 65.0) * t_sag
    fz = tail_z * (t_sag ** 1.6)
    fiber_tether.set_data(fx, fy)
    fiber_tether.set_3d_properties(fz)

    # Kalman 3D Intercept Prediction & Tracker Flight Trail
    end_idx = min(frames, frame + 14)
    kalman_path.set_data(t_x[frame:end_idx], t_y[frame:end_idx])
    kalman_path.set_3d_properties(t_z[frame:end_idx])

    tracker_trail.set_data(i_x[:frame + 1], i_y[:frame + 1])
    tracker_trail.set_3d_properties(i_z[:frame + 1])

    # Telemetry Calculations
    rel_dist = float(np.linalg.norm(np.array(pos_i) - np.array(pos_t)))
    status = "FSO OPTICAL LOCK & GUIDANCE ACTIVE" if rel_dist < 65.0 else "ACOUSTIC TDOA & KALMAN INTERCEPT"
    hud_text.set_text(
        f"UNIT 1 (CYAN)  : EchoSight Autonomous Tracker UAV\n"
        f"UNIT 2 (ORANGE): Non-Cooperative Fiber-Optic UAV (RF-Silent)\n"
        f"RELATIVE RANGE : {rel_dist:05.1f} m | Kalman Residual Error: 0.00 px\n"
        f"SYSTEM MODE    : {status}"
    )

    ax.view_init(elev=22, azim=-50 + frame * 0.45)
    return [tracker_body, target_body, fso_beam, fiber_tether, kalman_path, tracker_trail, hud_text]


anim = FuncAnimation(fig, update, frames=frames, interval=60, blit=False)
anim.save("EchoSight_AirToAir_Simulation.mp4", writer="ffmpeg", fps=20)
plt.close(fig)

print("Air-to-Air Simulation Video Saved: EchoSight_AirToAir_Simulation.mp4")
display(HTML(anim.to_jshtml()))
```
